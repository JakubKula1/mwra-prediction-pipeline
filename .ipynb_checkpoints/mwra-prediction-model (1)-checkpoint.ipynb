```python
def visualize_outliers(df, columns):
    # Set the style of seaborn
    sns.set(style="whitegrid")

    # Loop over columns to visualize
    for column in columns:
        # Create a new figure for each column
        plt.figure(figsize=(18, 5))

        # Box Plot
        plt.subplot(1, 3, 1)
        sns.boxplot(y=df[column])
        plt.title(f'Box Plot of {column}')
        plt.ylabel(column)

        # Scatter Plot
        plt.subplot(1, 3, 2)
        plt.scatter(df.index, df[column])
        plt.title(f'Scatter Plot of {column}')
        plt.ylabel(column)
        plt.xlabel('Index')

        # Histogram
        plt.subplot(1, 3, 3)
        sns.histplot(df[column], bins=30, kde=True)
        plt.title(f'Histogram of {column}')
        plt.xlabel(column)
        plt.ylabel('Frequency')

        # Ensure layout doesn't overlap and plots are fully displayed
        plt.tight_layout()
        plt.show()  # Ensure plot rendering
```

```python
def identify_outliers(df, columns):
    outliers_info = {}    
    for column_name in columns:
        column_data = df[column_name]
       
        lower_bound = column_data.quantile(0.25) - 1.5 * stats.iqr(column_data)
        upper_bound = column_data.quantile(0.75)+ 1.5 * stats.iqr(column_data)

        # Identify outliers
        identified_outlier = column_data[(column_data < lower_bound) | (column_data > upper_bound)] 
        outliers_info[column_name] = identified_outlier.count()

    return outliers_info

def remove_outliers(df, columns):
    for column_name in columns:
        column_data = df[column_name]
       
        lower_bound = column_data.quantile(0.25) - 1.5 * stats.iqr(column_data)
        upper_bound = column_data.quantile(0.75)+ 1.5 * stats.iqr(column_data)

        identified_outlier = column_data[(column_data < lower_bound) | (column_data > upper_bound)]
        df = df.drop(identified_outlier.index)

    return df
```

```python
columns_to_check = cln_connections.columns[3:]
visualize_outliers(cln_connections,columns_to_check)
```

```python
outliers_info = identify_outliers(cln_connections, columns_to_check)
pd.DataFrame(list(outliers_info.items()), columns=['Column Name', 'Outlier Count'])
```

**Graphs after removing outliers**

```python
cln_connections = remove_outliers(cln_connections,columns_to_check[:7])
visualize_outliers(cln_connections,columns_to_check[:7])
```

```python
def parse_date(date_str: str):
    date_formats = [
        '%m/%d/%Y, %H:%M:%S',  # Format: 05/24/2018, 00:00:00
        '%Y/%m/%d',             # Format: 2018/12/10
        '%d %b %Y',             # Format: 13 Mar 2024
        '%Y-%m-%d',             # Format: 2022-09-09
        '%Y/%m/%d',             # Format: 2023/10/22
        '%Y/%m/%d',             # Format: 2019/07/26
        '%m/%d/%Y, %H:%M:%S',   # Format: 08/19/2019, 00:00:00
        '%Y/%m/%d',             # Format: 2019/09/10
        '%m/%d/%Y, %H:%M:%S',   # Format: 03/07/2020, 00:00:00
        '%d %b %Y'              # Format: 28 Oct 2018
    ]
    
    for fmt in date_formats:
        try:
            return datetime.strptime(date_str, fmt)
        except ValueError:
            pass
    
    raise ValueError(f'Invalid format: {date_str}')
    
# inspired by #https://www.bhojkarki.com.np/Converting-Multiple-DateFormat-into-uniform-date-in-python/
```

```python
cln_profiles['parsed_date'] = cln_profiles['registration'].apply(parse_date)
cln_profiles = cln_profiles.drop(['registration'], axis=1)
cln_profiles['registration'] = cln_profiles['parsed_date'].dt.strftime('%Y-%m-%d')
cln_profiles = cln_profiles.drop(['parsed_date'], axis=1)
```

### 1.3 Formulation and Statistical Verification of Hypotheses

#### 1.3.1 Hypotheses

##### 1.3.1.1 Hypothesis I
**$H_0$** = _c.katana_ has, on average, a **higher** weight in the malware-related-activity state than in the normal state

```python
malware_state = connections[connections['mwra'] == 1]['c.katana']
normal_state = connections[connections['mwra'] == 0]['c.katana']
```

##### Q-Q plot

```python
plt.figure(figsize=(12, 6))
plt.suptitle('Q-Q Plots of mwra state and normal state', fontsize=16)

plt.subplot(1, 2, 1)
stats.probplot(malware_state, dist="norm", plot=plt)
plt.title('Mwra state')

plt.subplot(1, 2, 2)
stats.probplot(normal_state, dist="norm", plot=plt)
plt.title('Normal state')

plt.show()
```

##### Kolmogorov-Smirnov Test

```python
ks_stat, p_value = stats.ks_2samp(malware_state, normal_state)

print(f'Kolmogorov-Smirnov Test Statistic = {ks_stat}')
print(f'p-value = {p_value}')

alpha = 0.05
if p_value > alpha:
    print('- >Fail to reject H0')
else:
    print('-> Reject H0')
```

##### Levene Test

```python
l_stat, p_value = levene(malware_state, normal_state)

print(f'Levene Test Statistic = {l_stat}')
print(f'p-value = {p_value}')

alpha = 0.05
if p_value > alpha:
    print('-> Fail to reject H0')
else:
    print('-> Reject H0')
```

##### Mann-Whitney U Test

```python
mwu_stat, p_value = mannwhitneyu(malware_state, normal_state)

print(f'Mann-Whitney U Test Statistic = {mwu_stat}')
print(f'p-value = {p_value}')

alpha = 0.05
if p_value > alpha:
    print('-> Fail to reject H0')
else:
    print('-> Reject H0')
```

##### 1.3.1.2 Hypothesis II

c.android.youtube has a higher average weight in the normal state than in the malware-related activity.

```python
df = connections

normal_state = df[df['mwra'] == 0]['c.android.youtube']
malware_state = df[df['mwra'] == 1]['c.android.youtube']
plt.figure(figsize=(12, 6))
plt.suptitle('Q-Q Plots of mwra state and normal state', fontsize=16)

plt.subplot(1, 2, 1)
stats.probplot(malware_state, dist="norm", plot=plt)
plt.title('Mwra state')

plt.subplot(1, 2, 2)
stats.probplot(normal_state, dist="norm", plot=plt)
plt.title('Normal State')

plt.tight_layout()
plt.show()


```

QQ Plot looks **light-tailed** - in the tails there are some deviations from the line, even though data in the middle seems to follow normal distribution -> hard to decide just by looking at the graph, we need to do **Kolmogorov-smirnov**

**Kolmogorov-smirnov test**

**For the Normal State (mwra = 0):**
- **$H_0$ (null hypothesis)** The dataset `normal_state` follows a normal distribution.  
- **$H_1 = H_A$ (alternative hypothesis)**: The dataset `normal_state` does not follow a normal distribution.  

**For the Malware State (mwra = 1):**
- **$H_0$ (null hypothesis)**  The dataset `malware_state` follows a normal distribution.  
- **$H_1 = H_A$ (alternative hypothesis)**: The dataset `malware_state` does not follow a normal distribution.  

**P-Value** < 0.05: reject the null hypothesis. <br>
**P-Value** ≥ 0.05: doesnt reject the null hypothesis.

```python
ks_normal, p_normal = stats.kstest(normal_state, 'norm', args=(normal_state.mean(), normal_state.std()))
ks_malware, p_malware = stats.kstest(malware_state, 'norm', args=(malware_state.mean(), malware_state.std()))

print(f"Kolmogorov-Smirnov Test for Normal State: Statistic = {ks_normal}, P-Value = {p_normal}")
print(f"Kolmogorov-Smirnov Test for Malware State: Statistic = {ks_malware}, P-Value = {p_malware}")

if p_normal < 0.05:
    print("Normal State: Reject the null - not normal.")
else:
    print("Normal State: Fail to reject the null - likely normal.")

if p_malware < 0.05:
    print("Malware State: Reject the null - not normal.")
else:
    print("Malware State: Fail to reject the null - likely normal.")


```

**Levene test**

- **$H_0$ (null hypothesis)**: The variances of the dataset normal_state and malware_state are equal.
- **$H_1 = H_A$ (alternative hypothesis)**: The variances of the dataset normal_state and malware_state are not equal.

**P-Value** < 0.05: reject the null hypothesis (the variances are significantly different) <br>
**P-Value** ≥ 0.05: (the variances can be assumed to be equal).


```python
levene_statistic, p_levene = stats.levene(normal_state, malware_state)
print(f"Levene's Test: Statistic = {levene_statistic}, P-Value = {p_levene}")
if p_levene < 0.05:
    print("Levene's Test: Reject the null - variances differ.")
else:
    print("Levene's Test: Fail to reject the null - variances likely similar.")
```

**Mann-Whitney U Test**

Since data are not from normal distribution and variances are different we will use Mann-Whitney U Test.
c.android.youtube has a higher average weight in the normal state than in the malware-related activity.
- **$H_0$ (null hypothesis)**: The average weight of c.android.youtube in the normal state is less than or equal to the average weight in the malware-related activity state.
- **$H_1 = H_A$ (alternative hypothesis)**: The average weight of c.android.youtube in the normal state is greater than the average weight in the malware-related activity state.


```python
u_statistic, p_value = stats.mannwhitneyu(normal_state, malware_state, alternative='greater')
print(f"Mann-Whitney U Test: U Statistic = {u_statistic}, P-Value = {p_value}")

if p_value < 0.05:
    print("Reject the null hypothesis: Evidence supports that the average weight in the normal state is greater.")
else:
    print("Fail to reject the null hypothesis: No evidence that the average weight in the normal state is greater.")
```

#### 1.3.2 Statistical Power (Data Support) of the Tests Check

**Hypothesis I - statistical power**

```python
malware_state = connections[connections['mwra'] == 1]['c.katana']
normal_state = connections[connections['mwra'] == 0]['c.katana']
alpha = 0.05  
desired_power = 0.80  

mean_normal = normal_state.mean()
mean_malware = malware_state.mean()
std_normal = normal_state.std(ddof=1)
std_malware = malware_state.std(ddof=1)

pooled_std = np.sqrt(((len(normal_state) - 1) * std_normal**2 + (len(malware_state) - 1) * std_malware**2) / (len(normal_state) + len(malware_state) - 2))
effect_size = (mean_normal - mean_malware) / pooled_std

required_n = sm.stats.tt_ind_solve_power(effect_size=effect_size, alpha=alpha, power=desired_power)

print(f"Cohen's d (Effect Size): {effect_size:.3f}")
print(f"Required sample size for each group : {np.ceil(required_n):.0f}")
```

**Hypothesis II - - statistical power**

```python
df = connections

normal_state = df[df['mwra'] == 0]['c.android.youtube']
malware_state = df[df['mwra'] == 1]['c.android.youtube']
alpha = 0.05  
desired_power = 0.80  

mean_normal = normal_state.mean()
mean_malware = malware_state.mean()
std_normal = normal_state.std(ddof=1)
std_malware = malware_state.std(ddof=1)

pooled_std = np.sqrt(((len(normal_state) - 1) * std_normal**2 + (len(malware_state) - 1) * std_malware**2) / (len(normal_state) + len(malware_state) - 2))
effect_size = (mean_normal - mean_malware) / pooled_std

required_n = sm.stats.tt_ind_solve_power(effect_size=effect_size, alpha=alpha, power=desired_power)

print(f"Cohen's d (Effect Size): {effect_size:.3f}")
print(f"Required sample size for each group : {np.ceil(required_n):.0f}")
```
